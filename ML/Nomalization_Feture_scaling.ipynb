```python
import pandas as pd
import matplotlib.pylab as plt
import seaborn as sns
import numpy as np
```

```python
df=pd.read_csv("student_scaling_dataset.csv")
```

```python
df
```

```python
df.shape
```

```python
x=df.iloc[:,[0,1]] 
x
```

```python
y=df.iloc[:,[2]]
y
```

```python
df.describe()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test= train_test_split(x,y,test_size=0.3,random_state=40)
```

```python
len(x_train)
```

```python
len(y_test)
```

```python
x_test
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler=MinMaxScaler()
```

```python
x_train_scaler=scaler.fit_transform(x_train)
```

```python
x_train_scaler=pd.DataFrame(x_train_scaler,columns=x_train.columns)
x_train_scaler
```

```python
x_test_scaler=scaler.transform(x_test)
```

```python
x_test_scaler=pd.DataFrame(x_test_scaler,columns=x_test.columns)
```

```python
x_test_scaler
```

```python
np.round(x_train_scaler.describe(),1)
```

```python
np.round(x_train.describe(),1)
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,figsize=(15,5))
ax1.scatter(x_train["Study_Hours"],x_train["Attendance"],c=y_train.values)
ax2.scatter(x_train_scaler["Study_Hours"],x_train_scaler["Attendance"],c=y_train.values)
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,figsize=(15,5))
sns.kdeplot(x_train["Study_Hours"],ax=ax1,legend="Study")
sns.kdeplot(x_train["Attendance"],ax=ax1,legend="Attendance")

sns.kdeplot(x_train_scaler["Study_Hours"],ax=ax2,legend="Study")
sns.kdeplot(x_train_scaler["Attendance"],ax=ax2,legend="Attendance")
```
