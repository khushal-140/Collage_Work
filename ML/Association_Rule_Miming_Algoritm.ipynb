```python
import pandas as pd
import matplotlib.pylab as plt
```

```python
data= {
    "Tracation_ID":[1,2,3,4,5],
    "Item_Purchased":[["bread","milk"],["bread","milk","butter"],["bread","butter"],["milk","butter"],["bread","milk","butter"]]
}
```

```python
df=pd.DataFrame(data)
df
```

To Solve Asscoiation Algorithm we want to calclute 
i)Frequrnt Items Set
ii) Association rule genration

Now as we see then the one column is in text or string formate so 
we want to use encoder to encoder it but note but to it's is tracation dataset 
we want work with python (mixtend) libery import TaanscationEncode 


```python
from mlxtend.preprocessing import TransactionEncoder
te=TransactionEncoder()
df_encode=te.fit_transform(df["Item_Purchased"])
df_encode
```

```python
#convert into DataFrame
df_encode=pd.DataFrame(df_encode,columns=te.columns_)
df_encode
```

#Now impoert mlxtren Librery

```python
from mlxtend.frequent_patterns import association_rules,apriori
```

```python
frq_item_set=apriori(df_encode,min_support=0.4,use_colnames=True)
```

```python
rules=association_rules(frq_item_set,min_threshold=0.6)
```
