```python
import pandas as pd
import matplotlib.pylab as plt
```

```python
df=pd.read_csv("customers.csv")
```

```python
df
```

```python
df1=df.iloc[:,[3,4]]
df1
```

```python
#Feature Scaling 
from sklearn.preprocessing import MinMaxScaler
ms=MinMaxScaler()
df_scaler=ms.fit_transform(df1)
df_scaler
```

```python
df_scaler=pd.DataFrame(df_scaler,columns=df1.columns)
df_scaler
```

```python
#First Understand how many clustre are from using Dendrogram
import scipy.cluster.hierarchy as sch
plt.figure(figsize=(10,7))
den=sch.dendrogram(sch.linkage(df_scaler,method="ward"))
```

```python
#After Dendrogarm Implementation of Agglometive Clustring 
from sklearn.cluster import AgglomerativeClustering
cluter=AgglomerativeClustering(n_clusters=3)
label=cluter.fit_predict(df_scaler)
label
```
