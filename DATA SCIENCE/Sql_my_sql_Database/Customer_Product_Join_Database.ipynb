```python
import mysql.connector
```

```python
conn=mysql.connector.connect(
    host="localhost",
    username="root",
    password="",
    database="data_science"
)
print("Connect")
```

```python
cur=conn.cursor()
```

```python
cur.execute("""
        Create table if not exists Product
        (
        product_id varchar(10) primary key,
        product_name varchar(30) not null,
        price decimal(10,2)check(price>0)
        );
    """
)
conn.commit()
```

```python
cur.execute(
    """
        Create table if not exists Customer(
        customer_id varchar(10) primary key,
        customer_name varchar(30) not null,
        city varchar(30),
        product_id varchar(10),
        foreign key(product_id) references Product(product_id)
        );
    """
)
conn.commit()
```

```python
sql="""insert into product values (%s,%s,%s)"""
insert_product=[('P001', 'Laptop', 55000),
               ('P002', 'Smartphone', 25000),
               ('P003', 'Headphones', 2000),
               ('P004', 'Keyboard', 1500),
               ('P005', 'Monitor', 12000)]
cur.executemany(sql,insert_product)
conn.commit()
```

```python
# cur.execute("select c.customer_name,p.product_name from Customer c join Product p on c.product_id=p.product_id")

cur.execute("""
SELECT c.customer_name  ,p.product_name
FROM Customer c
 JOIN Product p
ON c.product_id = p.product_id;
""")
print("---------------Join Query--------------------")
print("Customer_name||Product_name")
join = cur.fetchall()
for row in join:
    print(row)
```

```python
cur.execute("""
    select c.customer_id,c.customer_name,product_name from 
    Customer c Left Join Product p
    ON c.product_id=p.product_id
    """)
left_join=cur.fetchall()
left_join
```

```python
cur.execute("""
    select c.customer_id,c.customer_name,product_name from 
    Customer c Right Join Product p
    ON c.product_id=p.product_id
    """)
right_join=cur.fetchall()
right_join
```
